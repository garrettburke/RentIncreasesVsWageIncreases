## Part 2: Data Preprocessing

### Section 2.1: Loading the Datasets

#### Subsection 2.1a: File Loading Functions

```python
def get_file_ext(file_path):
    return os.path.splitext(file_path)[1]


def load_file(file_path, header=True):
    file_type = get_file_ext(file_path)
    if file_type == '.csv':
        return pd.read_csv(file_path)
    if file_type in ('.xls', '.xlsx'):
        return pd.read_excel(file_path)


def load_rental_data(year):
    try:
        datasets[year]['rental_data'] = load_file(datasets[year]['rents'])
    except FileNotFoundError:
        print(f"Rental data file not found for the year {year}")

# needed a separate method for wage data because there are multiple wage
# data files for some years
def load_wage_data(year):
    try:
        wage_dfs = []
        for idx, file in enumerate(datasets[year]['wages']):
            if idx == 0:
                df = load_file(file)
            else:
                df = load_file(file, header=False)
            wage_dfs.append(df)
        datasets[year]['wage_data'] = pd.concat(wage_dfs, ignore_index=True)
    except FileNotFoundError:
        print(f"Wage data file not found for the year {year}")


def load_rental_and_wage_data(datasets):
    num_years = len(datasets)
    total_steps = 0

    for year in datasets:
        total_steps += 1
        total_steps += len(datasets[year]['wages'])

    with tqdm(total=total_steps, desc="Loading Datasets ... ", unit="step") as progress_bar:

        for year in datasets:
            load_rental_data(year)
            progress_bar.update(1)

            load_wage_data(year)
            progress_bar.update(len(datasets[year]['wages']))
```

### Section 2.2: Data Exploration

#### Subsection 2.2a: Dataset Inspection Functions

These features will save time inspecting the data, since there are so many files.

```python
def dataset_is_dataframe(dataset):
    return isinstance(dataset, pd.DataFrame)


def inspect_datasets(datasets, dataset_type):
    for year in datasets:
        if dataset_is_dataframe(datasets[year][dataset_type]):
            run_inspections(datasets[year][dataset_type], year, dataset_type)


def run_inspections(dataset, year, dataset_type):
    print('\n', str(year), dataset_type, '\n')
    print('Shape:', dataset.shape, '\n')
    dataset.info()
```

#### Subsection 2.3a: Creating CSV of CBSA and HUD_AREA_CODE Changes

After working on this project for a while, I've learned that CBSA numbers and hud_area_codes can change. There are some consistency issues with CBSA numbers between the population, rental, and wage datasets. This must be handled to ensure referential integrity is maintained. I will store old and new CBSA numbers and hud_area_codes in a CSV.

##### Loading CBSA and HUD Changes

```python
cbsa_changes = pd.read_csv('./data/cbsa_hud_changes.csv')
```

```python
cbsa_changes.info()
```

These columns are self-explanatory; METRO_NAME is included to make investigating these changes easier and will not be used for other purposes.

##### Functions to Update CBSA and HUD_AREA_CODE Numbers

```python
# .combine_first - uses CBSA_NEW if CBSA_NEW isn't null in the row
# else it takes the existing CBSA. Since I merged "right on the old CBSA num
# the new CBSA (if applicable) will not be null
def add_cbsa_column(df):
    df['cbsa'] = df['CBSA_NEW'].combine_first(df['cbsa'])
    return df


def drop_cbsa_and_suffix_cols(df):
    df = df.drop(columns=df.columns[df.columns.str.contains(
        'CBSA_', case=False, na=False)])
    df = df.drop(
        columns=df.columns[df.columns.str.contains('_x', case=False, na=False)])
    df = df.drop(
        columns=df.columns[df.columns.str.contains('_y', case=False, na=False)])
    return df


def update_cbsa_and_hud_numbers(df, cbsa_changes_df, data_type):
    if data_type == 'wage_data':
        old_num = 'WAGE_CBSA_OLD'
        new_num = 'WAGE_CBSA_NEW'
        left_on = 'cbsa'
        right_on = old_num
        cbsa_cols = [old_num, new_num]
        update_col = 'cbsa'
    if data_type == 'pop_data':
        old_num = 'POP_CBSA_OLD'
        new_num = 'POP_CBSA_NEW'
        left_on = 'cbsa'
        right_on = old_num
        cbsa_cols = [old_num, new_num]
        update_col = 'cbsa'
    if data_type == 'rental_data':
        old_num = 'hud_area_code_old'
        new_num = 'hud_area_code_new'
        left_on = 'hud_area_code'
        right_on = old_num
        cbsa_cols = [old_num, new_num]
        update_col = 'hud_area_code'
    if data_type == 'rental_data_final':
        old_num = 'RENT_CBSA_OLD'
        new_num = 'FINAL_CBSA_NEW'
        left_on = 'cbsa'
        right_on = old_num
        cbsa_cols = [old_num, new_num]
        update_col = 'cbsa'
    df = df.merge(
        cbsa_changes_df[cbsa_cols],
        how='left',
        left_on=left_on,
        right_on=right_on
    )
    if update_col is not None:
        df[update_col] = df[update_col].where(
            df[update_col] != df[old_num], df[new_num])
        return df
    else:
        raise Exception("Update column wasn't set - something went wrong.")
```

##### Function to Standardize Rental and Wage Datasets

```python
def standardize_data(corrected_col_names, dataset_type, filtered_cols):
    standardized_data = []
    for year in datasets:
        data = datasets[year].get(dataset_type)
        if dataset_is_dataframe(data):
            data.columns = data.columns.str.lower()
            data = data.dropna(how='all')
            data['year'] = year
            data = data.rename(columns=corrected_col_names)
            data = data[filtered_cols]
            standardized_data.append(data)
    return standardized_data
```

##### Solving the Problem: Rolling Up Metros in Old Data into the New CBSA

I have determined the best solution for these situations is to roll up the old data into a single CBSA. I'll combine the wage data for each CBSA into a single row. tot_emp can be used to create a weighted average for both the "mean" and "median" statistics.

Technically, this is a data transformation step. However, this must be done before combining with the population and rent datasets.

##### Wage Data Rollup Functions

```python
def roll_up_weighted_averages(filtered_df, cols, cbsa):
    total_employees = filtered_df['tot_emp'].sum()

    result = {}
    for col in cols:
        if total_employees > 0:
            result[col] = (filtered_df[col] * filtered_df['tot_emp']).sum() / total_employees
        else:
            result[col] = None

    return pd.Series(result, index=cols)
    

def rollup_tot_emp(df):
    return df['tot_emp'].sum()

    
def rollup_wages_by_cbsa(df, weighted_avg_cols):
    result = {}

    for (year, cbsa) in df.groupby(['year', 'cbsa']).groups.keys():
        subset = df[(df['year'] == year) & (df['cbsa'] == cbsa)]

        avg_series = roll_up_weighted_averages(subset, weighted_avg_cols, cbsa)

        total_emp = rollup_tot_emp(subset)

        combined_row = {'year': year, 'cbsa': cbsa, 'tot_emp': total_emp}

        combined_row.update(avg_series.to_dict())

        result[(year, cbsa)] = pd.Series(combined_row).to_frame().T

    return pd.concat(result.values(), ignore_index=True)
```

### Section 2.4: Data Transformation

#### Subsection 2.4a: Extracting States from Original Rental Data

State might be a valuable location feature. The problem with using the state from the "state_alpha" column included in the rental datasets was duplication. A metro area was sometimes included multiple times with a different state in the "state_alpha" column. I will create a function to extract all states from the rental dataset's state_alpha column and aggregate them into a single column. Then, I'll split that column on the space character into multiple state columns.

##### Function to Extract States by hud_area_code

```python
def extract_hud_area_state_mapping(datasets):
    combined_df = pd.DataFrame()

    for year in sorted(datasets.keys(), reverse=True):
        rental_data = datasets[year].get('rental_data')

        if dataset_is_dataframe(rental_data):
            renamed = rental_data.rename(columns=updated_rental_col_names)
            if 'hud_area_code' in renamed.columns and 'state_alpha' in renamed.columns:
                temp_df = renamed[['hud_area_code', 'state_alpha']]
                combined_df = pd.concat(
                    [combined_df, temp_df], ignore_index=True)
    grouped = (
        combined_df
        .groupby('hud_area_code')
        .agg({'state_alpha': lambda x: ' '.join(x.unique())})
        .reset_index()
    )

    return grouped.drop_duplicates()
```

```python
extracted_state_data = extract_hud_area_state_mapping(
    datasets).reset_index(drop=True)
```

##### Lagged Wage Features
The lagged wage features are lagged versions of the annual and hourly wage amounts.

##### Function to Create Lagged Version of Wage Features

The function looks for the column in the dataframe passed as df with a matching prefix and concatenates _lagged onto the column name, simultaneously creating a new column and assigning the lagged value of the column by using .shift(1).

```python
def lag_wage_prefix(df, prefix, years):
    for year in years:
        col_name = f"{prefix}"
        if col_name in df.columns:
            df[col_name + "_lagged"] = df.groupby("hud_area_code")[col_name].shift(1)
    return df
```

#### Subsection 2.4c: Transforming Data into Long Format

The next 2 steps are part feature engineering, part transformation.

##### Function to Get "Lagged" Value of Column Based on Year in Row

```python
def get_current_and_lagged_values(row, prefix):
    current_year_column = f'{prefix}{row["year"]}' # get row of current year
    prev_year = row['year'] - 1  # get row of previous year
    prev_year_column = f'{prefix}{prev_year}' # get column with population data prefix + prev year
    
    # get values for current year and previous year
    current_value = row[current_year_column] if current_year_column in row else None
    lagged_value = row[prev_year_column] if prev_year_column in row else None

    # returns a tuple (current_value, lagged_value)
    return current_value, lagged_value
```

##### Transforming BIRTHSXXXX Features to Long Format and Adding "Births Lagged" Feature in Each Row

This creates two new columns (births and births_lagged) in the all_data_lrw DataFrame. .apply() returns a series of tuples (1 tuple for each row returned from the get_current_and_lagged_values method) after applying the get_current_and_lagged_values function to each row of the all_data_lrw DataFrame. * unpacks the series of tuples. This will look like [(current_value1, lagged_value1), (current_value2, lagged_value2) ...], so on and so forth. Finally, getting to zip, which returns a zip object. The zip object iterates over tuples until none are left. The first item of each tuple will be assigned to the births column, and the second item to the births_lagged column.

By using this process, BIRTHSXXXX, DEATHSXXXX, POPESTIMATEXXXX, and NPOPCHGXXXX are transformed from wide to long format as well.

```python
all_data_lrw['births'], all_data_lrw['births_lagged'] = zip(*all_data_lrw.apply(get_current_and_lagged_values, axis=1, prefix='births'))
all_data_lrw['deaths'], all_data_lrw['deaths_lagged'] = zip(*all_data_lrw.apply(get_current_and_lagged_values, axis=1, prefix='deaths'))
all_data_lrw['pop_est'], all_data_lrw['pop_est_lagged'] = zip(*all_data_lrw.apply(get_current_and_lagged_values, axis=1, prefix='popestimate'))
all_data_lrw['pop_chg'], all_data_lrw['pop_chg_lagged'] = zip(*all_data_lrw.apply(get_current_and_lagged_values, axis=1, prefix='npopchg'))
```

```python
always_keep = ["lagged_rent_2"]
common_cols = [
    col for col in melted_rents_wl.columns
    if col in melted_lagged_rents_2.columns and col not in join_keys + always_keep
]
melted_lagged_rents_reduced = melted_lagged_rents_2.drop(columns=common_cols)
```

##### Creating List of Wide Format Columns to Remove

```python
lagged_rent_50_cols = [
    col for col in melted_rents_wl.columns if "lagged_rent_50" in col
]
```

##### Merging Melted Rents (Lag 1) with Melted Rents (Lag 2)

```python
melted_rents_wl2 = (
    pd.merge(melted_rents_wl, melted_lagged_rents_reduced, how="left", on=join_keys)
    .drop(columns=lagged_rent_50_cols)
    .drop_duplicates()
)
```

##### Verifying Bedroom Size, Lagged Rent, and Rent are in Long Format

```python
melted_rents_wl2.info()
```

#### Subsection 2.4d: More Feature Engineering

Several features I'd like to add require the data to be transformed into long format. Since that has been completed, now I will create the new features.

##### Creating "First Year Rent Amount" Feature

This feature gives the model a rent "floor" for each hud_area_code, since rent generally does not go down.

```python
min_year_per_hud = (
    melted_rents_wl2.groupby('hud_area_code')
    .agg({'year': 'min'})
    .reset_index()
    .rename(columns={'year': 'first_data_year'})
)

melted_rents_c = pd.merge(
    melted_rents_wl2,
    min_year_per_hud[['hud_area_code', 'first_data_year']],
    on='hud_area_code',
    how='left'
)

melted_rents_c = melted_rents_c.sort_values(
    ['hud_area_code', 'bedroom_size', 'year'])

first_year_rents = (
    melted_rents_c[melted_rents_c['year'] == melted_rents_c['first_data_year']][['hud_area_code', 'bedroom_size', 'rent']]
    .groupby(['hud_area_code', 'bedroom_size'])
    .first()
    .rename(columns={'rent': 'first_year_rent'})
    .reset_index()
)

melted_rents_c = melted_rents_c.merge(
    first_year_rents,
    on=['hud_area_code', 'bedroom_size'],
    how='left'
).drop_duplicates()

melted_rents_c['rent_change_since_first_year'] = (
    melted_rents_c['lagged_rent'] - melted_rents_c['first_year_rent']
)

melted_rents_c['rent_percent_change_since_first_year'] = (
    melted_rents_c['rent_change_since_first_year'] /
    melted_rents_c['first_year_rent']
) * 100

melted_rents_c['is_first_data_year'] = (
    melted_rents_c['year'] == melted_rents_c['first_data_year'].astype(int)
)

melted_rents_c['years_since_2000'] = melted_rents_c['year'] - 2000
```

##### Creating Rent Percentile Feature

```python
melted_rents_c['lagged_rent_percentile'] = (melted_rents_c.groupby(
    ['year', 'bedroom_size'])['lagged_rent'].rank(pct=True))
```

##### Creating  Wage to Lagged Rent Ratios

```python
melted_rents_c['median_wage_rent_ratio'] = melted_rents_c['a_median'] / (melted_rents_c['lagged_rent'])
melted_rents_c['mean_wage_rent_ratio'] = melted_rents_c['a_mean'] / (melted_rents_c['lagged_rent'])
```

##### Creating Total Area Wages to Lagged Rent Ratio

```python
melted_rents_c["median_wage_x_emp_rent_ratio"] = (
    melted_rents_c["a_median"]
    * melted_rents_c["tot_emp"]
    / (melted_rents_c["lagged_rent"])
)
melted_rents_c["mean_wage_x_emp_rent_ratio"] = (
    melted_rents_c["a_mean"]
    * melted_rents_c["tot_emp"]
    / (melted_rents_c["lagged_rent"])
)
```

#### Subsection 2.4e: Categorizing Feature Types

##### Setting Categories for Categorical, Nominal, Numerical and Ordinal Features

```python
nominal_features = ["state_1", "state_2", "state_3", "state_4", "hud_area_code", "cbsa", "hud_area_name"]
numerical_features = [
    "a_mean",
    "a_pct10",
    "a_pct25",
    "a_median",
    "a_pct75",
    "a_pct90",
    "h_mean",
    "h_pct10",
    "h_pct25",
    "h_median",
    "h_pct75",
    "h_pct90",
    "a_mean_lagged",
    "a_pct10_lagged",
    "a_pct25_lagged",
    "a_median_lagged",
    "a_pct75_lagged",
    "a_pct90_lagged",
    "h_mean_lagged",
    "h_pct10_lagged",
    "h_pct25_lagged",
    "h_median_lagged",
    "h_pct75_lagged",
    "h_pct90_lagged",
    "tot_emp",
    "births_lagged",
    "deaths_lagged",
    "pop_chg_lagged",
    "pop_est_lagged",
    "births",
    "deaths",
    "pop_est",
    "pop_chg",
    "tot_emp_lagged",
    "nchg_a_pct10",
    "nchg_a_pct25",
    "nchg_a_mean",
    "nchg_a_median",
    "nchg_a_pct75",
    "nchg_a_pct90",
    "nchg_h_pct10",
    "nchg_h_pct25",
    "nchg_h_median",
    "nchg_h_pct75",
    "nchg_h_pct90",
    "nchg_h_mean",
    "lagged_rent",
    "lagged_rent_2",
    "first_year_rent",
    "rent_change_since_first_year",
    "rent_percent_change_since_first_year",
    "year",
    "lagged_rent_percentile",
    "median_wage_rent_ratio",
    "mean_wage_rent_ratio",
    "mean_wage_x_emp_rent_ratio",
    "median_wage_x_emp_rent_ratio",
    "years_since_2000",
]

ordinal_features = ["bedroom_size"]

categorical_features = nominal_features + ordinal_features
all_features = nominal_features + numerical_features + ordinal_features
```

## Part 3: The Model

### Section 3.1: Selecting a Model

#### Subsection 3.1a: HistGradientBoostingRegressor

I selected this model because it’s meant for usage with time series data, and tree-based models can handle more features (high dimensionality) than other models. Two crucial features in the datasets are high cardinality features. “hud_area_code” represents a specific metro area, and “CBSA” represents a greater metro area. HistGradientBoostingRegressor includes native handling of categorical features, but unfortunately, it’s capped at 255 categories per feature. Instead, I had to use OneHotEncoder, which further increases dimensionality, supporting my choice of using a Tree-Based Model.

#### Subsection 3.1b: Importing Necessary Libraries and Setting Random State

```python
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.metrics import mean_absolute_percentage_error
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import TimeSeriesSplit
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder
```

### Section 3.2: Training the Model

#### Subsection 3.2a: Preparing to Train the Model

##### Setting Random State

Setting random state ensures results are reproducible. This controls the randomness in sklearn's random number generator.

```python
random_state = 42
```

##### Sorting the Data by Year

Since this is time-series data, it must be sorted in chronological order.

```python
melted_rents_c = melted_rents_c.sort_values(
    ['year', 'hud_area_code', 'bedroom_size']).reset_index(drop=True)
```

##### Defining X (Features) and y (Target)

Rent will be the target (y) since the goal is to predict rent prices. All other columns will be features (X). 2006 is the first year of data - I'm filtering this out since first_year_rent would be the rent amount from 2006.

```python
X = melted_rents_c[all_features]
y = melted_rents_c['rent']
```

##### Outputting X to CSV for Debugging Purposes

This allows me to investigate exactly what features/values are being sent to train the model.

```python
X.to_csv('X_train.csv')
```

#### Subsection 3.2b: Setting up Transformers

HistGradientBoostingRegressor isn't very sensitive to outliers, so passthrough can be used for the numerical transformer. In many cases, using OneHotEncoder isn't necessary with HistGradientBoostingRegressor because it handles categorical data natively. The problem is that this is capped at 255 unique values, and the CBSA and hud_area_code features are well above the cap. Since these features are above the cap, I need to use OneHotEncoder for the nominal features. My only ordinal feature is bedroom_size, and I'm using OrdinalEncoder, although I could skip this since the values are already [0, 1, 2, 3, 4]. In case I decide to add more ordinal features, I'll leave it as is.

```python
numerical_transformer = 'passthrough'
nominal_transformer = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
ordinal_transformer = OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=np.nan)
```

#### Subsection 3.2c: Setting up the Preprocessor

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numerical_transformer, numerical_features),
        ('nom', nominal_transformer, nominal_features),
        ('ord', ordinal_transformer, ordinal_features),
    ])
```

#### Subsection 3.2d: Setting up a Pipeline

```python
pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "regressor",
            HistGradientBoostingRegressor(
                early_stopping=True,
                l2_regularization=0.5,
                learning_rate=0.05,
                max_iter=500,
                max_leaf_nodes=31,
                min_samples_leaf=5,
                random_state=random_state,
                scoring="neg_mean_squared_error",
                validation_fraction=0.1
            ),
        ),
    ]
)
```

#### Subsection 3.2e: Time Series Split

TimeSeriesSplit automatically splits up the data into training and test splits, only training on past data and testing on future data. This prevents leaking future rent values while training.

```python
tscv = TimeSeriesSplit(n_splits=7)
```

#### Subsection 3.2f: Preparing DataFrame for Test Prediction Output and Creating RMSE Score List

```python
results_df = pd.DataFrame(columns=['year', 'hud_area_code', 'bedroom_size', 'rent_true', 'rent_pred'])
```

```python
pct_scores = []
rmse_scores = []
residuals = []
```

#### Subsection 3.2g: Training the Model and Predicting Rent on Test Data

```python
from sklearn.base import clone

for train_index, test_index in tscv.split(X):
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]
    
    pipeline.fit(X_train, y_train)

    y_pred = pipeline.predict(X_test)

    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    rmse_scores.append(rmse)
    mape = mean_absolute_percentage_error(y_test, y_pred)
    pct_error = mape * 100.0
    pct_scores.append(pct_error)
    
    residuals.extend(np.abs((y_test - y_pred) / y_test))
    
    temp_df = pd.DataFrame(
        {
            "year": melted_rents_c.iloc[test_index]["year"],
            "hud_area_code": melted_rents_c.iloc[test_index]["hud_area_code"],
            "bedroom_size": melted_rents_c.iloc[test_index]["bedroom_size"],
            "rent_true": y_test,
            "rent_pred": y_pred,
        }
    )
    results_df = pd.concat([results_df, temp_df], ignore_index=True)

results_df.to_csv("test_predictions.csv", index=False)

confidence_range = 0.90
q_pct = np.quantile(residuals, confidence_range)
```

#### Subsection 3.2h: Checking the Model's Accuracy on Test Data

```python
print(f"{confidence_range:.0%} of ***past*** predictions fall inside the range of ±{q_pct:.3%}")

avg_rmse = np.mean(rmse_scores)
print(f"rmse: {avg_rmse:.2f}")
avg_pct_error = np.mean(pct_scores)
print(f"pct_error: {avg_pct_error:.2f}")
```

#### Subsection 3.2i: Retraining the Model on the Entire Dataset

In subsection 3.2g, I trained and tested the model, but the model has not been trained on the entire dataset since I used TimeSeriesSplit. Below, I will clone the pipeline to get rid of anything the model might retain from the previous runs, and then I'll retrain the model on all data. Without doing this step, the predictions on the future data would likely be much less accurate.

```python
final_model = clone(pipeline)
final_model.fit(X, y)
```

## Part 4: Predicting Rent Values for Future Years

### Section 4.1: Preparing Estimates of Feature Values

#### Subsection 4.1a: Calculating Mean YoY Changes for Future Calculations

In order to use the model to predict, values are required for each of the features. The most straightforward way I've found to do this is simply using the average year-over-year change for each hud_area_code and multiplying the current year's value by 1 + YoY_Change to get next year's value.

##### Creating a List of the Original Numeric Features

These are all the numeric features from population and wage data, aside from births and deaths, which need to be handled separately. All of these features need values generated.

```python
original_num_features = [
    "a_mean",
    "a_pct10",
    "a_pct25",
    "a_median",
    "a_pct75",
    "a_pct90",
    "h_mean",
    "h_pct10",
    "h_pct25",
    "h_median",
    "h_pct75",
    "h_pct90",
    "pop_est",
    "tot_emp",
]
```

##### Looping Over the Numeric Features to Get YoY Change % for each hud_area_code

```python
agg_results = {}

for feature in original_num_features:
    df = melted_rents_c[["hud_area_code", "hud_area_name", "year", feature]].drop_duplicates()
    df = df.sort_values(by=["hud_area_code", "year"])

    df["prev"] = df.groupby("hud_area_code")[feature].shift(1)
    df[f"{feature}_yoy_pct"] = (df[feature] - df["prev"]) / df["prev"]

    df = df.dropna(subset=[f"{feature}_yoy_pct"])
    mean_changes = df.groupby("hud_area_code")[f"{feature}_yoy_pct"].mean()

    agg_results[f"{feature}_yoy_pct"] = mean_changes
```

##### Calculating Mean Births and Deaths Per Capita for each hud_area_code

```python
# I am using births/deaths per capita because applying YoY changes isn't appropriate for these features
births_df = melted_rents_c[["hud_area_code", "year", "births", "pop_est"]]
births_df = births_df.dropna()
births_df["births_per_capita"] = births_df["births"] / births_df["pop_est"]
births_pc_mean = births_df.groupby("hud_area_code")["births_per_capita"].mean()
agg_results["avg_births_per_capita"] = births_pc_mean

deaths_df = melted_rents_c[["hud_area_code", "year", "deaths", "pop_est"]]
deaths_df = deaths_df.dropna()
deaths_df["deaths_per_capita"] = deaths_df["deaths"] / deaths_df["pop_est"]
deaths_pc_mean = deaths_df.groupby("hud_area_code")["deaths_per_capita"].mean()
agg_results["avg_deaths_per_capita"] = deaths_pc_mean
```

##### Converting the Aggregated Results to a DataFrame

Below, I will convert the agg_results dict into a DataFrame to prepare it for usage in calculations of features which still need to be created. 

```python
agg_df = pd.DataFrame(agg_results).reset_index()
```

#### Subsection 4.1b: Creating a Dataframe with Estimated Feature Values

##### Setting Up a Baseline - 2024 Data

My plan is to use the feature values from the final year (2024) data as the baseline for estimating future feature values.

```python
final_year_values = melted_rents_c[melted_rents_c['year'] == 2024]
```

##### Create DataFrame with All Permutations of hud_area_code and Future Years (2025-2035)

I need to create a DataFrame for the forecasting which includes all future years, and a row for each hud_area_code for each of those future years.

```python
forecast_years = list(range(2024, 2036))
forecast_df = pd.DataFrame(
    [
        (code, year)
        for code in melted_rents_c["hud_area_code"].unique()
        for year in forecast_years
    ],
    columns=["hud_area_code", "year"],
)
```

##### Combining Forecast Years DataFrame (Blank Stats) with 2024 Data

Now I'll combine the 2024 data into the dataframe I just created. This gives me a dataframe with baseline values to extrapolate from, and a row for each hud_area_code in each future year.

```python
forecast_df = pd.merge(
    forecast_df, final_year_values, how="left", on=["hud_area_code", "year"]
)
```

##### Eliminating Anomalies from the Data

Below I'll select any hud_area_codes which don't have an a_mean statistic. This means no wage statistics are available for that hud_area_code. These need to be removed from the dataframe before prediction, so I'll use hud_anomalies with a negative filter.

```python
hud_anomalies = (forecast_df[(forecast_df['year'] == 2024) & (
    forecast_df['a_mean'].isna())]['hud_area_code'])
forecast_df = forecast_df[~forecast_df['hud_area_code'].isin(hud_anomalies)]
```

##### Adding YoY Changes to forecast_df

Now that I have a dataframe with all the future years and rows for each hud_area_code within those years, I can merge it together with agg_df, which is the dataframe I created earlier containing mean YoY changes.

```python
for feature in original_num_features:
    yoy_col = feature
    forecast_df = pd.merge(
        forecast_df,
        agg_df[["hud_area_code", f"{yoy_col}_yoy_pct"]],
        on="hud_area_code",
        how="left",
    )
```

##### Adding YoY Growth to Base Stats in forecast_df for Future Years

Having the mean YoY changes makes estimating the future values of features straightforward. Here is the formula for calculating compound growth:\
baseline value * (1 + mean YoY change) ^ number of years since 2024.

```python
def apply_growth(row, yoy_col, baseline_2024):
    base = baseline_2024.get(row['hud_area_code'], np.nan)
    if pd.isna(base):
        return np.nan
    return base * (1 + row[yoy_col]) ** row['years_since_2024']


forecast_df['years_since_2024'] = forecast_df['year'] - 2024

for feature in original_num_features:
    yoy_col = f'{feature}_yoy_pct'

    forecast_df[yoy_col] = forecast_df[yoy_col].fillna(0)

    baseline_2024 = (
        forecast_df[forecast_df['year'] == 2024]
        .set_index(['hud_area_code'])[feature]
        .to_dict()
    )

    forecast_df[feature] = forecast_df.apply(
        apply_growth,
        axis=1,
        args=(yoy_col, baseline_2024)
    )
```

##### Setting Values for Lagged Wage Features

The values of the lagged wage features are currently set to NaN for future years. Since the future wage feature values have been estimated, the lagged values can now be set simply by using .shift(1) to get the value from the prior year. 

```python
for feature in original_num_features:
    forecast_df[f"{feature}_lagged"] = forecast_df.groupby("hud_area_code")[
        feature
    ].shift(1)
```

##### Updating Net Change Features

The net change features are currently set to NaN for future years and need to be updated. Since the compound growth formula was applied to these features, the net changes can now be calculated.

```python
for feature in original_num_features:
    lagged = forecast_df.groupby("hud_area_code")[feature].shift(1)
    net_change = forecast_df[feature] - lagged

    # I don't want to overwrite my 2024 data, so I am specifying only years after 2024
    forecast_df.loc[forecast_df["year"] > 2024, f"nchg_{feature}"] = net_change

forecast_df["pop_chg"] = forecast_df["pop_est"] - forecast_df["pop_est_lagged"]
forecast_df["pop_chg_lagged"] = forecast_df["pop_chg"].shift(1)
```

##### Calculating and Adding Births and Deaths Features

Births and Deaths need to be calculated still. I will do this by using the avg_births_per_capita and avg_deaths_per_capita that were calculated earlier in agg_df. Then, I will create births_lagged and deaths_lagged.

```python
forecast_df = pd.merge(
    forecast_df,
    agg_df[["hud_area_code", "avg_births_per_capita", "avg_deaths_per_capita"]],
    on="hud_area_code",
    how="left",
)

forecast_df["births"] = forecast_df["pop_est"] * forecast_df["avg_births_per_capita"]
forecast_df["deaths"] = forecast_df["pop_est"] * forecast_df["avg_deaths_per_capita"]
forecast_df["births_lagged"] = forecast_df.groupby("hud_area_code")["births"].shift(1)
forecast_df["deaths_lagged"] = forecast_df.groupby("hud_area_code")["deaths"].shift(1)
```

##### Splitting the Dataset Into Future and Past Sets

The future years in forecast_df only contain a single row per hud_area_code, which means the permutations for bedroom_size don't exist. This must be fixed. First, I will remove the 2024 data because it already has all the bedroom sizes.

```python
forecast_df_2024 = forecast_df[forecast_df["year"] == 2024]
forecast_df = forecast_df[forecast_df["year"] > 2024]
```

##### Adding Permutations for Bedroom Sizes to Future Data

Below, I will remove the bedroom_size column from the filtered forecast_df to prevent duplication after the merge, create the bedroom_sizes dataframe, and merge back to forecast_df using cross join. Using a cross join gives each bedroom size to every row in forecast_df - which is exactly what I want.

```python
forecast_df = forecast_df.drop(columns="bedroom_size", errors="ignore")
bedroom_sizes = pd.DataFrame({"bedroom_size": range(5)})
forecast_df = forecast_df.merge(bedroom_sizes, how="cross")
```

##### Getting First Year Rent Values from 2024 Data

The first-year rent feature values are still empty in the future dataframe (forecast_df). This can be retrieved from the 2024 data.

```python
first_year_data = forecast_df_2024[["hud_area_code", "bedroom_size", "first_year_rent"]].groupby(["hud_area_code", "bedroom_size"]).first()
forecast_df = forecast_df.drop(columns=['first_year_rent']).merge(first_year_data, on=["hud_area_code", "bedroom_size"], how="left")
```

##### Setting the "Years Since 2000" Feature and "Is First Data Year Y/N" Feature

```python
forecast_df["years_since_2000"] = forecast_df["year"] - 2000
forecast_df["is_first_data_year"] = (forecast_df["year"] == forecast_df["first_data_year"]).astype(int)
```

##### Merging the 2024 Data Back with the Future Data

Since the rent values from 2024 will be used as a starting point for calculating lagged rent, I need to merge the 2024 and future dataframes back together.

```python
forecast_df = pd.concat([forecast_df, forecast_df_2024])
```

##### Adding Static Features to Future Years

The last features the model needs are the static features: CBSA, states, and the first data year. After that, I can move on to prediction.

```python
static_features = [
    "cbsa",
    "state_1",
    "state_2",
    "state_3",
    "state_4",
    "first_data_year",
    "hud_area_name"
]

static_2024 = (
    forecast_df[forecast_df["year"] == 2024]
    .drop_duplicates(subset=["hud_area_code"])[["hud_area_code"] + static_features]
    .set_index("hud_area_code")
)

for feature in static_features:
    forecast_df[feature] = forecast_df["hud_area_code"].map(static_2024[feature])
```

### Section 4.2: Using the Model to Predict Future Rent Values

#### Subsection 4.2a: Running the Model and Adding Calculated Features to Future Years

```python
forecast_df_copy = forecast_df.sort_values(["year", "hud_area_code", "bedroom_size"]).copy()

final_yearly = []

for year in forecast_years[1:]:
    forecast_df_copy["lagged_rent"] = forecast_df_copy.groupby(["hud_area_code", "bedroom_size"])["rent"].shift(1)
    forecast_df_copy["lagged_rent_2"] = forecast_df_copy.groupby(["hud_area_code", "bedroom_size"])["rent"].shift(2)

    current_year = forecast_df_copy["year"] == year

    forecast_df_copy.loc[current_year, "median_wage_x_emp_rent_ratio"] = (
        forecast_df_copy.loc[current_year, "a_median"]
        * forecast_df_copy.loc[current_year, "tot_emp"]
        / forecast_df_copy.loc[current_year, "lagged_rent"]
    )

    forecast_df_copy.loc[current_year, "mean_wage_x_emp_rent_ratio"] = (
        forecast_df_copy.loc[current_year, "a_mean"]
        * forecast_df_copy.loc[current_year, "tot_emp"]
        / forecast_df_copy.loc[current_year, "lagged_rent"]
    )

    forecast_df_copy.loc[current_year, "median_wage_times_lr"] = (
        forecast_df_copy.loc[current_year, "a_median"]
        * forecast_df_copy.loc[current_year, "lagged_rent"]
    )

    forecast_df_copy.loc[current_year, "median_wage_rent_ratio"] = (
        forecast_df_copy.loc[current_year, "a_median"]
        / forecast_df_copy.loc[current_year, "lagged_rent"]
    )

    forecast_df_copy.loc[current_year, "mean_wage_rent_ratio"] = (
        forecast_df_copy.loc[current_year, "a_mean"]
        / forecast_df_copy.loc[current_year, "lagged_rent"]
    )

    pct = (
        forecast_df_copy.loc[current_year]
        .groupby("bedroom_size")["lagged_rent"]
        .rank(pct=True, ascending=False)
    )

    forecast_df_copy.loc[current_year, "lagged_rent_percentile"] = pct

    forecast_df_copy.loc[current_year, "rent_change_since_first_year"] = (
        forecast_df_copy.loc[current_year, "lagged_rent"]
        - forecast_df_copy.loc[current_year, "first_year_rent"]
    )
    forecast_df_copy.loc[current_year, "rent_percent_change_since_first_year"] = (
        forecast_df_copy.loc[current_year, "rent_change_since_first_year"]
        / forecast_df_copy.loc[current_year, "first_year_rent"]
    ) * 100

    y_pred = final_model.predict(forecast_df_copy.loc[current_year, all_features])
    forecast_df_copy.loc[current_year, "rent"] = y_pred

    final_yearly.append(forecast_df_copy.loc[current_year])

final_results_df = pd.concat(final_yearly, ignore_index=True)
```

### Section 4.3: Heatmaps Function

My original plan was to create bar charts with two bars: rent and median wages, and draw a line across the bars to note 30%. I tried doing this, and it became clear that even looking at a single metro area, the results were small and difficult to read. Using heatmaps seems like a much better solution; all data points (year, bedroom size, and rent to 30% wage ratio) are clearly displayed and easily interpreted.

```python
# This function creates a heatmap for a single hud_area_code.
# The color of the cells is determined based on the ratio of
# rent divided by 30% of the monthly median wage. Creates a 
# directory for each CBSA and saves the heatmaps using the
# hud_area_code as the file name.
def plot_affordability_heatmap(df):
    hac = str(df["hud_area_code"].iloc[0])
    han = df["hud_area_name"].iloc[0]
    cbsa_name = df["name"].iloc[0]

    temp_df = df.copy()
    # Monthly rent to monthly median wage ratio
    temp_df["ratio"] = temp_df["rent"] / (temp_df["a_median"] / 12 * 0.30)

    # My bedroom_size column is currently in long format, e.g.
    # each row represents 1 bedroom size, but I need the data
    # to be in wide format, 1 column per bedroom size. The
    # pivot_table function is basically the opposite of .melt().
    # If the row in temp_df is bedroom_size = 0 it will take
    # the ratio value I created for bedroom_size = 0 and when
    # the 0 bedroom size column is created, that ratio will be
    # the value in the column for that years row. This must be
    # done to get the data into "wide" format for .imshow()
    pivoted_data = (
        temp_df.pivot_table(index="year", columns="bedroom_size", values="ratio")
        .sort_index()
        .values
    )
    
    years = sorted(temp_df["year"].unique())
    num_beds = sorted(temp_df["bedroom_size"].unique())

    fig, ax = plt.subplots(figsize=(8, 6))
    im = ax.imshow(pivoted_data, aspect="auto", cmap=plt.get_cmap("RdYlGn_r"), vmax=2, vmin=0)

    cbar = fig.colorbar(im, ax=ax)
    cbar.set_label("Rent to 30% of Median Wage Ratio")

    ax.set_xticks(np.arange(len(num_beds)))
    ax.set_xticklabels([f"{int(b)} BR" for b in num_beds])
    ax.set_yticks(np.arange(len(years)))
    ax.set_yticklabels(years)

    title = f"CBSA: {cbsa_name} \n"
    title += f"HUD Area Code: {hac}\n"
    title += f"Metro: {han}"
    ax.set_title(title, loc="left", multialignment="left")
    ax.set_xlabel("Bedroom Size")
    ax.set_ylabel("Year")
    
    plt.tight_layout()
    plt.show()
    dir = "heatmaps/" + cbsa_name.strip("/")
    filename = os.path.join(dir, f"{hac}.png")

    os.makedirs(dir, exist_ok=True)

    if os.path.isfile(filename):
        os.remove(filename)

    fig.savefig(filename, dpi=150)
    

# This function creates heatmaps for each hud_area_code within the specified CBSA
def make_cbsa_hud_heatmaps(df, cbsa_code):
    cbsa_df = df[df["cbsa"] == cbsa_code]
    cbsa_df = cbsa_df.merge(population_data[["cbsa", "name"]], on="cbsa")

    for hac, df in cbsa_df.groupby("hud_area_code"):
        try:
            plot_affordability_heatmap(df)
        except Exception as E:
            print(E)
            print(f"Error processing {hac} in CBSA {cbsa_code}")

# National version of plot_affordability_heatmap function
def plot_national_affordability_heatmap(df):
    temp_df = df.copy()
    temp_df["ratio"] = temp_df["rent"] / (temp_df["a_median"] / 12 * 0.30)

    pivoted_data = (
        temp_df.pivot_table(index="year", columns="bedroom_size", values="ratio", aggfunc="mean")
        .sort_index()
        .values
    )

    years = sorted(temp_df["year"].unique())
    num_beds = sorted(temp_df["bedroom_size"].unique())

    fig, ax = plt.subplots(figsize=(8, 6))
    im = ax.imshow(pivoted_data, aspect="auto", cmap=plt.get_cmap("RdYlGn_r"), vmax=2, vmin=0)

    cbar = fig.colorbar(im, ax=ax)
    cbar.set_label("Rent to 30% of Median Wage Ratio")

    ax.set_xticks(np.arange(len(num_beds)))
    ax.set_xticklabels([f"{int(b)} BR" for b in num_beds])
    ax.set_yticks(np.arange(len(years)))
    ax.set_yticklabels(years)

    title = "Nationwide Affordability: Median Rent to 30% of Median Wage Ratio"
    ax.set_title(title, loc="left")
    ax.set_xlabel("Bedroom Size")
    ax.set_ylabel("Year")

    plt.tight_layout()
    plt.show()

    dir = "heatmaps/national"
    filename = os.path.join(dir, "national_affordability.png")

    os.makedirs(dir, exist_ok=True)

    if os.path.isfile(filename):
        os.remove(filename)

    fig.savefig(filename, dpi=150)
```
